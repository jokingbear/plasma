```python
import plasma.parallel_processing as pp
import time
import threading
```

```python
connection = 'tcp://0.0.0.0:1212'
```

```python
inputs = pp.zmq.PubZMQ(connection, qsize=2)
inputs2 = pp.zmq.PubZMQ(connection, qsize=2)
```

```python
def wait_print(x):
    thread_id = threading.get_ident()
    print(f'thread_id:{thread_id} - value: {x}')
    time.sleep(1)
```

```python
for i in range(10):
    if i % 2 == 0:
        inputs.put(i)
    else:
        inputs2.put(i)
```

```python
outputs = pp.zmq.SubZMQ(connection, n=4).register_callback(wait_print).run()
```
