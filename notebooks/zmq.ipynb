```python
import plasma.parallel_processing as pp
import time
import threading
```

```python
connection = 'ipc:///tmp/test.ipc'
```

```python
inputs = pp.zmq.PubZMQ(connection, qsize=2, timeout=5)
```

```python
def wait_print(x):
    thread_id = threading.get_ident()
    print(f'thread_id:{thread_id} - value: {x}')
    time.sleep(1)
```

```python
outputs = pp.zmq.SubZMQ(connection, n=4).register_callback(wait_print).run()
```

```python
for i in range(10):
    # print(i)
    inputs.put(i)
```
