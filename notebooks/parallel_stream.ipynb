```python
import time
import plasma.parallel_processing as pp

from plasma.parallel_processing.stream import parallel
```

```python
pp.mp.set_start_method('fork')
```

```python
def wait(x:int):
    time.sleep(2)
    return x + 1

def print_pass(x:int):
    print(x)
    return x
```

```python
s = (
    parallel(range(10))
    .select_by_process(wait, num_worker=4)
    .select_by_thread(print_pass, num_worker=2)
    .stream()
)
s.evaluate()
```
